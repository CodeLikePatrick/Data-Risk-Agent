# Exploratory data analysisProfile the synthetic dataset before modelling: size, null rates, risk rate, and the price and title distributions.

```python
import pandas as pdfrom ecomrisk.data import generatefrom ecomrisk.engine import build_featuresmerchants, products = generate(seed=42)print('merchants', merchants.shape, 'products', products.shape)merchants.head()
```

```python
merchants.isna().sum().to_frame('nulls')
```

```python
merchants['true_risk'].value_counts().to_frame('count').assign(rate=lambda d: d['count'] / d['count'].sum())
```

```python
features = build_features(merchants, products)features.describe().T[['mean', 'std', 'min', '50%', 'max']]
```

```python
products.groupby('category')['price'].agg(['count', 'min', 'median', 'max', 'mean', 'std'])
```
